````python
from vllm import LLM, SamplingParams
LLM_MODEL = 'Qwen/Qwen3-4B-Instruct-2507'
llm = LLM(model=LLM_MODEL,dtype="float16",tensor_parallel_size=2,max_model_len=1024,gpu_memory_utilization=0.90)
````

````python
import itertools
import json
import random

# 1. Define Taxonomy
tech_labels = [
    "Artificial Intelligence",
    "Cloud Computing",
    "Cybersecurity",
    "Quantum Computing",
    "Data Science",
    "Telecommunications",
    "Software Engineering",
    "Autonomous Vehicles",
    "Space Technology",
    "Robotics",
    "Internet of Things (IoT)",
    "Financial Technology (FinTech)",
    "Augmented & Virtual Reality (AR/VR)",
    "Cryptography",
]
science_labels = [
    "Genomics",
    "Astrophysics",
    "Materials Science",
    "Neuroscience",
    "Physics",
    "Astrophysics",
    "Particle Physics",
    "Quantum Physics",
    "Nuclear Physics",
    "Chemistry",
    "Organic Chemistry",
    "Inorganic Chemistry",
    "Physical Chemistry",
    "Mathematics",
    "Statistics",
    "Applied Mathematics",
    "Aerospace",
    "Climate Science",
    "Energy Science",
]

# all_labels = tech_labels + science_labels

formats = [
    "Research Paper Abstract",
    "Tech Blog Post",
    "Industry News Brief",
    "Patent Application Summary",
    "Scientific News Article",
    "Technical Report",
    "Product Announcement",
    "Company Press Release",
    "Policy Brief",
    "Case Study",
    "Research Finding Report",
    "Technology Trend Report",
    "Science Explainer",
    "Technical Documentation",
    "Market Research Summary",
]
styles = [
    "Formal",
    "Technical",
    "Journalistic",
    "Academic",
    "Analytical",
    "Descriptive",
    "Concise",
    "Explanatory",
    "Business-oriented",
    "Critical",
]

lengths = ["Short (50-100 words)", "Medium (100-200 words)", "Long (200-350 words)"]
````

````python
prompt_template = """SYSTEM PROMPT:
You are an expert dataset generator specializing in creating high-quality, realistic synthetic text for machine learning models.

TASK:
Generate a single synthetic dataset item for a multi-label text classification model operating across the TECHNOLOGY and SCIENCE domains.

TARGET LABELS TO IMPLY:
- Selected Labels: {TARGET_LABELS}

STYLISTIC & DOMAIN CONSTRAINTS:
- Format / Genre: {TEXT_FORMAT} (e.g., peer-reviewed abstract excerpt, tech blog entry, research forum post, product announcement, patent abstract)
- Tone / Audience: {TONE_STYLE} (e.g., highly academic, practitioner-focused, journalistic, speculative)
- Length: {LENGTH_RANGE} words.
- Opening Sentence Style: {OPENING_STYLE}

GENERATION RULES:
1. Multi-Label Grounding: The generated text MUST genuinely embody ALL the topics listed in {TARGET_LABELS}. The topics must be naturally interwoven rather than forced.
2. Avoid Explicit Label Speller ("Label Leakage"): DO NOT simply repeat the exact name of the target labels in the text unless necessary. Express the underlying concepts, methodologies, and terminology instead.
3. Sentence Starter Variety: Strictly adhere to the requested Opening Sentence Style. NEVER start the text with repetitive filler openers like "A new...", "A recent...", "The project...", or "In this paper...".
4. Realistic Noise & Edge Cases: Make the text sound authentic to its format. Include relevant domain jargon, technical acronyms, or real-world contextual nuances.
5. Output Format: Return STRICTLY a valid JSON object. Do not include markdown code block syntax (e.g., ```json) or any explanatory text before or after the JSON.

EXPECTED JSON SCHEMA:
{{
  "text": "The generated synthetic paragraph...",
  "labels": ["Label1", "Label2"],
  "difficulty_level": "beginner | intermediate | advanced"
}}"""
````

````python
import re
opening_styles = [
    # Direct Action & Imperatives
    "Start with a direct action or gerund verb (e.g., 'Deploying...', 'Optimizing...')",
    "Start with an imperative instruction (e.g., 'Configure the cluster...', 'Ensure that...')",
    
    # Scenarios & Problem Framing
    "Start with a problem statement or scenario (e.g., 'When handling high traffic...')",
    "Start with a pain point or anti-pattern (e.g., 'A common mistake in microservices is...')",
    "Start with a hypothetical condition (e.g., 'If your primary database fails...')",
    
    # Direct Technical Naming & Definitions
    "Start with a technical concept or tool name directly (e.g., 'Kubernetes handles...', 'Redis caching...')",
    "Start with a crisp technical definition (e.g., 'Idempotency ensures that...')",
    
    # Time, Preconditions & Sequence
    "Start with a temporal or conditional phrase (e.g., 'Before configuring...', 'Prior to release...')",
    "Start with a step in a sequence (e.g., 'First, establish a baseline...')",
    
    # Industry Context & Comparisons
    "Start with a contrast or trade-off (e.g., 'Unlike relational databases, key-value stores...')",
    "Start with an industry standard or best practice (e.g., 'Modern CI/CD pipelines require...')",
    
    # Analytical & Question-Driven
    "Start with a technical question (e.g., 'Why do distributed systems require...')",
    "Start with a quantitative or metric-focused observation (e.g., 'Reducing latency by 20% requires...')"
]

def generate_prompt_payload():
    # Pick 2 to 3 random overlapping labels to enforce multi-label coverage
    k = random.choice([1,2,3,4,5,6])
    selected_labels = random.sample(tech_labels, k)

    fmt = random.choice(formats)
    tone = random.choice(styles)

    # Return parameters to format into the system prompt template above
    return {
        "TARGET_LABELS": selected_labels,
        "TEXT_FORMAT": fmt,
        "TONE_STYLE": tone,
        "LENGTH_RANGE": random.choice(lengths),
        "OPENING_STYLE": random.choice(opening_styles),
    }

def extract_json_payload(text: str):
    """Extracts JSON content using regular expressions to handle leading/trailing text."""
    # Search for content inside ```json ... ``` or standard ``` ... ```
    match = re.search(r"```(?:json)?\s*(\{.*\}|\[.*\])\s*```", text, re.DOTALL)
    if match:
        return match.group(1).strip()
    
    # Fallback search for bare structural JSON bounds
    match = re.search(r"(\{.*\}|\[.*\])", text, re.DOTALL)
    if match:
        return match.group(1).strip()
        
    return text.strip()
    
# 2. Batch Construction
NUM_SAMPLES = 20000  # Adjust as needed
BATCH_SIZE = 100
sampling_params = SamplingParams(
    temperature=0.8,
    top_p = 0.9,
    max_tokens=1024,
    stop=["<|im_end|>", "<|endoftext|>"]
)


dataset = []
failed_count = 0

for i in range(0, NUM_SAMPLES, BATCH_SIZE):
    chunk_size = min(BATCH_SIZE, NUM_SAMPLES - i)
    print(f"Processing batch {i // BATCH_SIZE + 1}/{(NUM_SAMPLES + BATCH_SIZE - 1) // BATCH_SIZE} ({chunk_size} items)...")
    
    batch_prompts = []
    batch_metadata = []
    
    for _ in range(chunk_size):
        payload = generate_prompt_payload()
        formatted_prompt = prompt_template.format(**payload)
        chat_prompt = f"<|im_start|>system\nYou are a helpful assistant. Output valid JSON only.<|im_end|>\n<|im_start|>user\n{formatted_prompt}<|im_end|>\n<|im_start|>assistant\n"
        
        batch_prompts.append(chat_prompt)
        batch_metadata.append(payload)
        
    outputs = llm.generate(batch_prompts, sampling_params, use_tqdm=True)
    
    for output, meta in zip(outputs, batch_metadata):
        raw_text = output.outputs[0].text
        cleaned_json_str = extract_json_payload(raw_text)

        try:
            parsed_json = json.loads(cleaned_json_str)
            dataset.append({
                "text": parsed_json.get("text", ""),
                "labels": parsed_json.get("labels", []),
                "difficulty_level": parsed_json.get("difficulty_level", "intermediate"),
                "target_labels_requested": meta["TARGET_LABELS"],
                "format": meta["TEXT_FORMAT"],
                "style": meta["TONE_STYLE"]
            })
        except json.JSONDecodeError:
            failed_count += 1
            continue

print(f"Dataset generated with {len(dataset)} valid rows. (Failed parse count: {failed_count})")
````

````python
import pandas as pd
df = pd.DataFrame(dataset)
df['all_labels'] = df.apply(lambda _: tech_labels, axis=1)
df
````

````python


parquet_filename = "synthetic_multilabel_dataset_tech.parquet"
df.to_parquet(parquet_filename, engine="pyarrow", index=False)

print(f"Successfully generated {len(df)} valid items and saved to {parquet_filename}")
````
